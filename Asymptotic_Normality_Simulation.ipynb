```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Parameters
p = 0.3  # True probability of success
n = 500  # Sample size per simulation
num_samples = 5000  # Number of samples/simulations to run
```

```python
# 2. Simulate 5,000 samples of size 500
# np.random.binomial counts successes out of n trials.
# Dividing by n converts the count into a sample proportion (p-hat).
sample_proportions = np.random.binomial(n, p, num_samples) / n
```

```python
# 3. Calculate and print the empirical vs. predicted Standard Deviation (SD)
empirical_sd = np.std(sample_proportions)
predicted_sd = np.sqrt(
    (p * (1 - p)) / n
)  # Formula from slide

print(f"Empirical SD: {empirical_sd:.4f}")
print(f"Predicted SD: {predicted_sd:.4f}")
```

```python
# 4. Plot the bell-shaped histogram centered at p = 0.3
plt.hist(
    sample_proportions, bins=30, density=True, edgecolor="black", alpha=0.7
)
plt.axvline(
    p, color="red", linestyle="dashed", linewidth=2, label=f"True Mean (p={p})"
)
plt.title(
    "Numerical Verification: Asymptotic Normality (Bernoulli, p=0.3, n=500)"
)
plt.xlabel("Sample Proportion (p-hat)")
plt.ylabel("Density")
plt.legend()
plt.show()
```
